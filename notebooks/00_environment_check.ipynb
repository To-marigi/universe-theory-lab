## 記号式から高速な数値関数へ

SymPyでラグランジアンから運動方程式を導出し、NumPy関数へ変換します。

```python
import sympy as sp

t, m, omega = sp.symbols('t m omega', positive=True, real=True)
q = sp.Function('q')(t)
L = m * sp.diff(q, t)**2 / 2 - m * omega**2 * q**2 / 2
eom = sp.simplify(sp.diff(sp.diff(L, sp.diff(q, t)), t) - sp.diff(L, q))
sp.Eq(eom, 0)
```
